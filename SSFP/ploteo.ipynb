```python

import json
import plotly.graph_objects as go
import plotly.io as pio
pio.renderers.default = "vscode"

def html_a_figura(path):
    """Recupera la figura de Plotly a partir de un .html guardado con write_html."""
    html = open(path, encoding="utf-8").read()
    i = html.rfind("Plotly.newPlot(")
    if i < 0:
        raise ValueError("no parece un HTML de Plotly")
    s = html[i + len("Plotly.newPlot("):].lstrip()
    dec = json.JSONDecoder()
    _, j = dec.raw_decode(s);      s = s[j:].lstrip(" ,\n")   # id del div (se descarta)
    data, j = dec.raw_decode(s);   s = s[j:].lstrip(" ,\n")   # trazas
    layout, _ = dec.raw_decode(s)                             # layout

    layout.pop("template", None)          # template de otra versión de Plotly: se descarta
    fig = go.Figure(data=data, layout=layout, skip_invalid=True)
    fig.update_layout(template="plotly_white")
    return fig

# path = r"C:\Users\Usuario\Desktop\5to Semestre\Aprendizaje profundo con vision artificial\ShipDeNet\ShipDeNet-20\SSFP\visualizacion_val.html-20261002T205011Z-1-001\visualizacion_val.html\curva_pr.html"
# fig = html_a_figura(path)
# fig.show()
```
